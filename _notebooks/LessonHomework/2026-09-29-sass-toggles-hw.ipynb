## Homework — 3 interactive toggles with live count + features

- 3 toggles, each with own label/purpose, all on `ocs__toggle` grammar.
- JS reads `.checked` to update a live count (`aria-live`) and drive real features: highlight section, show/hide study tips, compact mode.
- 4+ insight comments included.


```python
%%html
<!-- UI_RUNNER: Toggles Homework -->
<label class="ocs__toggle ocs__toggle--green">
  <input id="t-highlight" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Highlight key terms</span>
</label>

<label class="ocs__toggle ocs__toggle--green">
  <input id="t-tips" class="ocs__toggle-input" type="checkbox" checked>
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Show study tips</span>
</label>

<label class="ocs__toggle ocs__toggle--green">
  <input id="t-compact" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Compact mode</span>
</label>

<p><output id="toggle-count" aria-live="polite">2 of 3 toggles on</output></p>

<div id="demo-panel" class="ocs__card">
  <h3>Demo panel</h3>
  <p id="demo-text">Toggle the switches above — this panel highlights, collapses tips, and compacts.</p>
  <p id="demo-tips"><em>Tip:</em> .checked is a live boolean — flip a switch and the count + panel update instantly.</p>
</div>

<script>
// I learned .checked lets JS read a toggle as a true/false condition for selection.
const toggles = document.querySelectorAll('.ocs__toggle-input');
const count = document.getElementById('toggle-count');
const panel = document.getElementById('demo-panel');
const tips = document.getElementById('demo-tips');

// Each toggle drives a different feature branch — one listener updates everything.
function updateToggles() {
  const on = [...toggles].filter(t => t.checked).length;
  count.textContent = `${on} of ${toggles.length} toggles on`;
  // Feature 1: highlight proves a toggle can restyle another element.
  panel.style.background = document.getElementById('t-highlight').checked ? '#fef9c3' : '';
  // Feature 2: show/hide proves a toggle can control visibility.
  tips.hidden = !document.getElementById('t-tips').checked;
  // Feature 3: density proves a toggle can switch layout modes.
  panel.style.padding = document.getElementById('t-compact').checked ? '4px' : '';
}
toggles.forEach(t => t.addEventListener('change', updateToggles));
updateToggles();
</script>
```

The example below earns the extra credit: one toggle drives dark mode across the whole panel, another reveals a hidden image, and the comments explain the reasoning behind each line. Copy the pattern into your own page to submit for the higher rubric band.


```python
%%html
<!-- UI_RUNNER: Toggles Extra Credit -->
<label class="ocs__toggle ocs__toggle--green">
  <input id="ec-dark" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Dark mode</span>
</label>

<label class="ocs__toggle ocs__toggle--green">
  <input id="ec-image" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Show surprise image</span>
</label>

<p><output id="ec-status" aria-live="polite">0 of 2 extra-credit toggles on</output></p>

<section id="ec-panel" class="ocs__card">
  <h3>Extra credit panel</h3>
  <p>This whole section reacts to the toggles above — proof the switch controls a real feature.</p>
  <img id="ec-img" alt="Surprise" width="160"
       src="https://pages.opencodingsociety.com/images/logo.png" hidden>
</section>

<script>
// Insight: a toggle is just selection — its .checked boolean is the condition,
// and the branches below choose which feature runs.
const dark = document.getElementById('ec-dark');
const imageToggle = document.getElementById('ec-image');
const panel = document.getElementById('ec-panel');
const image = document.getElementById('ec-img');
const status = document.getElementById('ec-status');

// Insight: reading .checked inside the handler (not once at load) keeps the UI
// in sync — the boolean is re-evaluated every time a toggle fires "change".
function renderExtraCredit() {
  // Feature 1: dark mode restyles the whole panel, not just the switch itself.
  panel.style.background = dark.checked ? '#111827' : '';
  panel.style.color = dark.checked ? '#f9fafb' : '';

  // Feature 2: .hidden mirrors the boolean directly — no CSS class bookkeeping.
  image.hidden = !imageToggle.checked;

  // Insight: aria-live="polite" makes screen readers announce this count change,
  // so the feature stays accessible without extra JavaScript.
  const on = [dark, imageToggle].filter(t => t.checked).length;
  status.textContent = on + ' of 2 extra-credit toggles on';
}

[dark, imageToggle].forEach(t => t.addEventListener('change', renderExtraCredit));
renderExtraCredit();
</script>
```
