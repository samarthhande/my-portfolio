
## Homework Hack (independent): SFI Backend Validator

Build a reusable validator for candidate SFI automotive records.

A valid record has:

- a non-empty product name,
- an accepted racing category (`Auto Racing` or `Drag Racing`),
- a non-empty spec number,
- a non-empty effective date,
- and a spec number that does **not** already exist in the backend.

The provided starter data includes three cases. The completed runner below validates every supplied record, prints ACCEPT or REJECT, and also tests edge cases: an empty product name, an invalid category, and a duplicate spec. The validator is self-contained so it can run on its own.


**Output recorded:** the three supplied records produce `ACCEPT RECORD` for spec `1.2`, `REJECT RECORD` for spec `3.1` (invalid `Street Car` category), and `REJECT RECORD` for spec `1.1` (duplicate). The edge cases also reject an empty product name, an invalid category, and a duplicate spec.

**Explanation:** the validator breaks the English rule into named Boolean checks before combining them. `and` enforces all required fields, `in valid_categories` implements the `Auto Racing OR Drag Racing` rule, and `not duplicate_spec` rejects specs already stored. Returning the `checks` dictionary makes each accept/reject decision traceable back to the exact condition that failed.


### College Board pseudocode trace

The same rule in AP CSP pseudocode:

```
hasProductName <- (productName != "")
validCategory <- (category = "Auto Racing") OR (category = "Drag Racing")
hasSpecNumber <- (specNumber != "")
hasEffectiveDate <- (effectiveDate != "")
duplicateSpec <- (specNumber = "1.1") OR (specNumber = "2.1")
isValid <- hasProductName AND validCategory AND hasSpecNumber AND hasEffectiveDate AND NOT duplicateSpec

IF (isValid)
{
    DISPLAY("ACCEPT RECORD")
}
ELSE
{
    DISPLAY("REJECT RECORD")
}
```

Trace: `1.2 / Auto Racing / new` -> `T AND T AND T AND T AND NOT F` = `T` -> ACCEPT. `3.1 / Street Car` -> `validCategory = F` -> REJECT. `1.1 / Drag Racing / duplicate` -> `NOT T = F` -> REJECT. The Python validator makes the same decisions.
