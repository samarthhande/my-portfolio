## Popcorn Hack 2 — Flask route

Change: custom roster names + extra `/coach` route (procedure with a web address). `jsonify` turns the dict into JSON.


````python
# CODE_RUNNER: Popcorn 2 - Flask scoreboard routes
from flask import Flask, jsonify

app = Flask(__name__)

# My roster (changed per task)
@app.route("/score")
def score():
    return jsonify({"home": 58, "away": 52, "quarter": 3})

@app.route("/roster")
def roster():
    return jsonify({"players": ["Samarth", "Ryden", "Aarav"]})

# My extra route (pattern copied): coach of the season stats app
@app.route("/coach")
def coach():
    return jsonify({"coach": "Coach Hande", "season": "2026-27"})

client = app.test_client()
print("/score says:", client.get("/score").get_json())
print("/roster says:", client.get("/roster").get_json())
print("/coach says:", client.get("/coach").get_json())
````

## Homework Hack — Season Stats Report (soccer)

Uses 5 stdlib modules: `json`, `datetime`, `statistics`, `random`, `math`. `summarize_season` has a full API docstring. Own sport (soccer goals), own games, next game 2027-03-01. `requirements.txt` lists only externals (`pandas`, `Flask`); stdlib is never listed.


````python
# CODE_RUNNER: Homework - Season Stats Report (soccer)
import json
import math
import random
import statistics
from datetime import date

games = [
    {"date": "2026-09-14", "opponent": "Lincoln", "goals": 2, "minutes": 80},
    {"date": "2026-09-15", "opponent": "Madison", "goals": 1, "minutes": 70},
    {"date": "2026-09-16", "opponent": "Lincoln", "goals": 3, "minutes": 85},
    {"date": "2026-09-18", "opponent": "Roosevelt", "goals": 0, "minutes": 60},
]

def summarize_season(season):
    """Summarize a list of soccer games.

    Parameters:
        season: list of dictionaries, each with "date", "opponent", "goals", and "minutes"
    Returns:
        dictionary with total_points, average_points, best_game,
        and points_by_opponent (goals work as points here)
    """
    points = [game["goals"] for game in season]
    points_by_opponent = {}
    for game in season:
        opponent = game["opponent"]
        points_by_opponent[opponent] = points_by_opponent.get(opponent, 0) + game["goals"]
    return {
        "total_points": sum(points),
        "average_points": round(statistics.mean(points), 1),
        "best_game": max(points),
        "points_by_opponent": points_by_opponent,
    }

report = summarize_season(games)

# datetime: days until my next game; math: round a save-percentage stat; random: pick a MOTM
next_game = date(2027, 3, 1)
report["days_until_next_game"] = (next_game - date.today()).days
report["save_pct_rounded_up"] = math.ceil(87.2)
report["motm_pick"] = random.choice([g["opponent"] for g in games])

print(json.dumps(report, indent=2))

# requirements.txt lists only external packages. json, datetime, statistics,
# math, and random come with Python, so they are never listed.
requirements_txt = """pandas  # season table
Flask  # scoreboard web routes
"""
print()
print("requirements.txt:")
print(requirements_txt)
````
