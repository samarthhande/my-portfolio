Rule: HTML for meaning, `ocs_` classes for role. No custom CSS, no inline styles, no `<style>` block.


## Popcorn Hack (in-class) — refactor to semantic HTML

Bad pattern was `<div class="title-text">` + `<p class="bold">` with custom CSS. Fixed below: one `h1` + one supporting subheading + one `p` with `ocs_lead`, semantic `em` for emphasis.


```python
%%html
<!-- UI_RUNNER: Typography Popcorn -->
<h1 class="ocs_section-title">All About Cats</h1>
<h2 class="ocs_badge">Why Gentle Handling Matters</h2>
<p class="ocs_lead">Never pull a cat's tail - <em>gentle</em> touches keep cats happy and calm.</p>
```

**What changed:** `div.title-text` → `h1.ocs_section-title`, plain `div` subheading → `h2.ocs_badge`, `p.bold` → `p.ocs_lead` with `<em>` instead of visual-only bold. Deleted the `<style>` block entirely; the SASS theme now styles it.


## Homework Hack — How to Pat a Cat refactor

Removed all 5 custom classes (`.title-font`, `.section-font`, `.body-text`, `.very-important`, `.list-item`) and the `<style>` block. Applied heading hierarchy (`h2` → `h3` → `h4`), paragraphs, one real `<ol>` with three `<li>` items, `<strong>` for the important line, and `ocs_card` + `ocs_section-title` theme classes.


```python
%%html
<!-- UI_RUNNER: Typography Homework Base -->
<div class="ocs_card">
<h2 class="ocs_section-title">How to Pat a Cat</h2>
<h3 class="ocs_badge">Have you ever wanted to...</h3>
<p class="ocs_text">Pat a cat? Well good for you — we will give step by step instructions on how to pat a cat!</p>
<p class="ocs_text"><strong>How to pat a cat:</strong></p>
<ol>
<li>Put out your hand</li>
<li>Touch the cat</li>
<li>Move your hand back and forth</li>
</ol>
<h4 class="ocs_status">Congrats! You pat a cat :D</h4>
</div>
```

### Validation checklist

- `%%html` + `UI_RUNNER` present: yes
- Custom classes / inline styles / `<style>` / `span`-emphasis: absent
- Headings + `p` + `ol`/`li`: present (h2/h3/h4, 2 paragraphs, 3 list items)
- `ocs_` classes: `ocs_card`, `ocs_section-title`, `ocs_badge`, `ocs_text`, `ocs_status`
- Semantic emphasis: `<strong>` + `<em>` (popcorn)
