## Popcorn Hack 2 — Size beakers with `fr`

Goal: 4 beakers in ONE row, Control Beaker twice as wide. `fr` count = boxes per row; `fr` size = width share.

`fr` values used = `1fr 1fr 1fr 2fr` (Control gets 2 of 5 shares).


````python
%%html
<!-- UI_RUNNER: Grids Popcorn 2 -->
<div class="ocs__grid" style="display: grid; grid-template-columns: 1fr 1fr 1fr 2fr; gap: 8px;">
<div class="ocs__grid-cell"><h3>Beaker 1</h3><p>pH 7.1</p></div>
<div class="ocs__grid-cell"><h3>Beaker 2</h3><p>pH 6.8</p></div>
<div class="ocs__grid-cell"><h3>Beaker 3</h3><p>pH 8.2</p></div>
<div class="ocs__grid-cell"><h3>Control Beaker</h3><p>pH 7.0 — twice as wide (2fr)</p></div>
</div>

<!-- Note: inline grid-template-columns is allowed ONLY for this fr-technique hack per lesson safety rules. -->
````
