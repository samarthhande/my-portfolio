# SASS Button Grammar HW — My Good Brain

Lesson: https://pages.opencodingsociety.com/sass/buttons

Rule: semantic tags (`<button>` = action, `<a>` = navigation) + OCS classes only (`ocs__btn`, `pill`, `fill`, `outline`, `accent`, `alert-green/yellow/red`, `small`). Groups in `ocs__links` / `ocs__links--wide`. No `<style>`, no inline styles, no hex colors.


## Popcorn Hack 1 — base button class


```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 1 -->
<a class="ocs__btn" href="/resources">Resource Directory</a>
```

## Popcorn Hack 2 — modifiers + group container

Added `pill` + `fill` to the check-in button; wrapped both in `ocs__links`.


```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 2 -->
<div class="ocs__links">
  <button class="ocs__btn pill accent fill">Complete Daily Check-In</button>
  <button class="ocs__btn outline">Save Draft Journal</button>
</div>
```

## Popcorn Hack 3 — tones + wide bar

Second link `alert-yellow` → `alert-green`; container `ocs__links` → `ocs__links--wide`; Crisis button gains `fill`.


```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 3 -->
<div class="ocs__links--wide">
  <button class="ocs__btn alert-green fill">Complete A Mental Health Check-In Form</button>
  <a class="ocs__btn alert-green outline" href="">Request Break</a>
  <a class="ocs__btn alert-red fill" href="">Get Crisis Support</a>
</div>
```

## Homework Hack 1 — Adaptive Resource Toolbar

Urgent = `alert-red fill pill`, Self-Care = `alert-green fill pill`, Community = `accent outline pill`, all in `ocs__links`. Inline styles + hex colors removed.


```python
%%html
<!-- UI_RUNNER: Buttons Homework 1 -->
<div class="ocs__links">
  <a href="https://www.mygoodbrain.org/resources" class="ocs__btn alert-red fill pill">Urgent Support</a>
  <a href="https://www.mygoodbrain.org/resources" class="ocs__btn alert-green fill pill">Self-Care</a>
  <a href="https://www.mygoodbrain.org/resources" class="ocs__btn accent outline pill">Community</a>
</div>
```

## Homework Hack 2 — Primary vs Secondary hierarchy

Primary `accent fill` > secondary `outline` > alert `alert-red outline`, all in `ocs__links`.


```python
%%html
<!-- UI_RUNNER: Buttons Homework 2 -->
<div class="ocs__links">
  <button class="ocs__btn accent fill">Log Mood Entry</button>
  <a href="https://www.mygoodbrain.org/resources" class="ocs__btn outline">View Weekly Trends</a>
  <a href="https://www.mygoodbrain.org/resources" class="ocs__btn alert-red outline">Crisis Hotline</a>
</div>
```
