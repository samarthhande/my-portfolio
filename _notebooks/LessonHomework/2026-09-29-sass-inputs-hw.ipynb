# SASS Inputs HW — OCS Input Grammar

Lesson: https://pages.opencodingsociety.com/sass/inputs/

Rule: `ocs__input` + modifiers (`small`, `medium`, `large`, `gradient`) only. No inline styles, no made-up classes. HTML creates the input, SASS styles it, JS reads `.value`, pseudocode uses `INPUT()`/`DISPLAY()`.


## Popcorn Hack — refactor two inputs

Expected: one large gradient input + one compact plain input, both on `ocs__input`.


```python
%%html
<!-- UI_RUNNER: Inputs Popcorn -->
<label for="pvo-search">Search PVO resources</label>
<input id="pvo-search" name="resourceRequest" type="text" class="ocs__input large gradient" placeholder="Search PVO resources...">

<label for="vol-name">Volunteer first name</label>
<input id="vol-name" name="volunteerName" type="text" class="ocs__input small" placeholder="Volunteer First Name">
```

## Homework Hack — signup form refactor

Removed `style` widths/backgrounds/padding and `.tinybox`. Sizes: Assistance Category = `large gradient` (decorative gradient needs the modifier), First Name = `small`, Email = `medium` + `type="email"`. Every field has a `<label>` for accessibility.


```python
%%html
<!-- UI_RUNNER: Inputs Homework -->
<form>
<label for="assist-cat">Assistance category</label>
<input id="assist-cat" name="assistanceCategory" type="text" class="ocs__input large gradient" placeholder="Assistance Category">

<label for="hw-first">Volunteer first name</label>
<input id="hw-first" name="volunteerName" type="text" class="ocs__input small" placeholder="Volunteer First Name">

<label for="hw-email">Volunteer email</label>
<input id="hw-email" name="volunteerEmail" type="email" class="ocs__input medium" placeholder="Volunteer Email">

<button type="submit" class="ocs__btn">Submit</button>
</form>

<script>
document.querySelector('form').addEventListener('submit', (e) => {
  e.preventDefault();
  const cat = document.getElementById('assist-cat').value;
  alert('Request received for: ' + (cat || '(empty — try typing a need)'));
});
</script>
```

### Validation checklist

- `%%html` + `UI_RUNNER`: present
- Inline `style` / made-up classes: absent
- `ocs__input` on every input: yes (3/3)
- Modifiers: `large`+`gradient`, `small`, `medium`: yes
- Accessibility: visible `<label>` + correct `type` + placeholder on each field
