```javascript
update() {
    if (this.revComplete) return;
    
    // Check lifespan for timed scythes
    if (this.lifespan && this.creationTime) {
        const currentTime = Date.now();
        if (currentTime - this.creationTime >= this.lifespan) {
            this.revComplete = true;
            this.destroy();
            return;
        }
    }
    
    // Custom motion update
    this.updateMotion();

}
```

```javascript
// GameLevelFortress.js - Keyboard event handling
setupInterceptorControls() {
    this.boundFireInterceptor = this.fireInterceptor.bind(this);
    document.addEventListener('keydown', this.boundFireInterceptor);
}

fireInterceptor(event) {
    if ((event.code === 'Space' || event.keyCode === 32) ||
        (event.code === 'KeyI' || event.keyCode === 73)) {
        event.preventDefault();
        // Create interceptor at player position
        const interceptor = new Interceptor(this.gameEnv, spawnX, spawnY);
        this.gameEnv.gameObjects.push(interceptor);
    }
}
```

// Define a class called Person
class Person {
    constructor(firstName, lastName, age) {
        this.firstName = firstName; // string property
        this.lastName = lastName;   // string property
        this.age = age;             // number property
    }

    // Method to return full name
    getFullName() {
        return this.firstName + " " + this.lastName;
    }

    // Method to check if the person is an adult
    isAdult() {
        return this.age >= 18;
    }

    // Method to celebrate birthday
    haveBirthday() {
        this.age += 1;
        console.log("Happy birthday, " + this.getFullName() + "! You are now " + this.age + " years old.");
    }
}

// Create instances of the Person class
let person1 = new Person("Alice", "Smith", 25);
let person2 = new Person("Bob", "Johnson", 16);

// Use methods
console.log(person1.getFullName()); // Alice Smith
console.log("Is adult?", person1.isAdult()); // true

console.log(person2.getFullName()); // Bob Johnson
console.log("Is adult?", person2.isAdult()); // false

// Celebrate birthday
person2.haveBirthday(); // Happy birthday, Bob Johnson! You are now 17 years old.
